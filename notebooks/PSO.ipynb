```python
# Import necessary libraries
import pyswarms as ps  # Library for implementing Particle Swarm Optimization
from sklearn.model_selection import train_test_split  # For splitting dataset
from sklearn.ensemble import RandomForestClassifier  # For classification model
from sklearn.metrics import accuracy_score  # For evaluating the model
from sklearn.datasets import load_iris  # Load sample Iris dataset
import numpy as np  # For numerical operations

# Load the Iris dataset
# X contains the features, and y contains the target labels
data = load_iris()
X, y = data.data, data.target

# Define the fitness function for PSO
def fitness_function(particle):
    """
    Calculate the fitness of each particle.
    Each particle represents a feature subset (binary vector).

    Parameters:
    - particle: A matrix where each row is a particle (binary vector).

    Returns:
    - scores: A numpy array of negative accuracy values (to minimize).
    """
    n_particles = particle.shape[0]  # Number of particles in the swarm
    scores = []  # List to store fitness scores for all particles

    for i in range(n_particles):
        # Identify selected features (indices where particle == 1)
        selected_features = np.where(particle[i] == 1)[0]

        # If no features are selected, assign a low fitness score
        if len(selected_features) == 0:
            scores.append(-1)  # Invalid solution
            continue

        # Subset the dataset to only include selected features
        X_selected = X[:, selected_features]

        # Split the dataset into training and validation sets
        X_train, X_val, y_train, y_val = train_test_split(
            X_selected, y, test_size=0.2, random_state=42
        )

        # Train a Random Forest classifier on the selected features
        model = RandomForestClassifier()
        model.fit(X_train, y_train)

        # Evaluate the model's accuracy on the validation set
        accuracy = accuracy_score(y_val, model.predict(X_val))

        # Append the negative accuracy as the fitness score (minimization problem)
        scores.append(accuracy)

    return np.array(scores) * -1  # Return negative scores to minimize

# Define PSO parameters
options = {
    'c1': 0.5,  # Cognitive parameter (particle's own experience)
    'c2': 0.3,  # Social parameter (swarm's experience)
    'w': 0.9 ,   # Inertia weight (controls exploration vs exploitation)
    'k': 3,     # Number of neighbors for topology
    'p': 2      # Use "lbest" (local best) topology
}

# Number of features in the dataset
dimensions = X.shape[1]

# Initialize the PSO optimizer
optimizer = ps.discrete.BinaryPSO(
    n_particles=10,       # Number of particles in the swarm
    dimensions=dimensions,  # Number of features (dimensionality)
    options=options         # Hyperparameters for PSO
)

# Perform the optimization
# Optimize the fitness function for a specified number of iterations
best_cost, best_pos = optimizer.optimize(fitness_function, iters=50)

# Display the results
print("Best Feature Subset:", np.where(best_pos == 1)[0])  # Indices of selected features
print("Best Fitness Score:", -best_cost)  # Best accuracy (negated back to positive)
```
