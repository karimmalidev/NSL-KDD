```python
import numpy as np
import pandas as pd
```

let's work on our training set first

```python
#introduse the path of the csv file
train_path=r'C:\Users\Gaming\Desktop\NSL-KDD Dataset In CSV Format\KDDTrain+.csv'

#creating the dataframe from pandas lib
data_frame = pd.read_csv(train_path, header=None)
```

```python
#printing the first five elements
print(data_frame.head())
```

```python
#list of headers that matter provided me with
headers = (['duration','protocol_type','service','flag','src_bytes','dst_bytes','land','wrong_fragment','urgent','hot','num_failed_logins','logged_in','num_compromised','root_shell','su_attempted','num_root','num_file_creations','num_shells','num_access_files','num_outbound_cmds','is_host_login','is_guest_login','count','srv_count','serror_rate','srv_serror_rate','rerror_rate','srv_rerror_rate','same_srv_rate','diff_srv_rate','srv_diff_host_rate','dst_host_count','dst_host_srv_count','dst_host_same_srv_rate','dst_host_diff_srv_rate','dst_host_same_src_port_rate','dst_host_srv_diff_host_rate','dst_host_serror_rate','dst_host_srv_serror_rate','dst_host_rerror_rate','dst_host_srv_rerror_rate','attack','level'])

# Assigning list of headers to the DataFrame's columns
data_frame.columns = headers

# Display the DataFrame with the new headers
print(data_frame.head())
```

```python
print(data_frame.info())
```

```python
print(data_frame.describe())
```

```python
# Create a mapping for 'normal' and others
data_frame['attack'] = data_frame['attack'].map({'normal': 0}).fillna(1)

# Drop the 'level' column
data_frame = data_frame.drop(columns=['level'])

# View the modified DataFrame
print(data_frame.head())
```

```python
# Rename the 'attack' column to 'label'
data_frame = data_frame.rename(columns={'attack': 'label'})

print(data_frame.info())
```

```python
#now we are gonna save the new data in a new CSV file 
data_frame.to_csv(r'C:\Users\Gaming\Desktop\NS_lab\lab1\modified_data_sets\KDD_Train_Set.csv', index=False)
```
